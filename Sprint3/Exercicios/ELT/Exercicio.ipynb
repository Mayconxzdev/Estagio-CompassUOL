```python
def ler_arquivo(filepath):
    with open(filepath, 'r', encoding='utf-8') as file:
        linhas = file.readlines()
    return linhas


def encontrar_ator_com_mais_filmes(linhas):
    max_filmes = 0
    ator_max_filmes = ""
    
    for linha in linhas[1:]: 
        colunas = linha.strip().split(',')
        
        if len(colunas) < 3: 
            continue  
        
        ator = colunas[0]
        try:
            num_filmes = int(colunas[2].strip())  
        except ValueError:
            continue  
        
        if num_filmes > max_filmes:
            max_filmes = num_filmes
            ator_max_filmes = ator
    
    return ator_max_filmes, max_filmes


def escrever_resultado(filepath, ator, num_filmes):
    with open(filepath, 'w', encoding='utf-8') as file:
        file.write(f"{ator} - {num_filmes}\n") 

# Caminhos dos arquivos
caminho_csv = 'actors.csv'
caminho_saida = 'etapa-1.txt'

linhas = ler_arquivo(caminho_csv)
ator, num_filmes = encontrar_ator_com_mais_filmes(linhas)
escrever_resultado(caminho_saida, ator, num_filmes)


```

```python
def calcular_media_gross(linhas):
    total_gross = 0
    count = 0
    
    for linha in linhas[1:]:  
        colunas = linha.strip().split(',')
        
        if len(colunas) < 6: 
            continue
        
        try:
            gross = float(colunas[5].strip())  
        except ValueError:
            continue 
            
        total_gross += gross
        count += 1
    

    media_gross = total_gross / count if count != 0 else 0
    return media_gross


def escrever_resultado(filepath, media_gross):
    with open(filepath, 'w', encoding='utf-8') as file:
        file.write(f"Média de receita bruta dos principais filmes: {media_gross:.2f} milhões de dólares\n")


caminho_csv = 'actors.csv'
caminho_saida = 'etapa-2.txt'


linhas = ler_arquivo(caminho_csv)
media_gross = calcular_media_gross(linhas)
escrever_resultado(caminho_saida, media_gross)

```

```python
def encontrar_ator_com_maior_media(linhas):
    max_media = 0
    ator_max_media = ""
    
    for linha in linhas[1:]: 
        colunas = linha.strip().split(',')
        
        if len(colunas) < 4:  
            continue
        
        ator = colunas[0]
        
        try:
            media = float(colunas[3].strip()) 
        except ValueError:
            continue  
        
        if media > max_media:
            max_media = media
            ator_max_media = ator
    
    return ator_max_media, max_media


def escrever_resultado(filepath, ator, media):
    with open(filepath, 'w', encoding='utf-8') as file:
        file.write(f"{ator} - {media:.2f} milhões de dólares por filme\n")


caminho_csv = 'actors.csv'
caminho_saida = 'etapa-3.txt'


linhas = ler_arquivo(caminho_csv)
ator, media = encontrar_ator_com_maior_media(linhas)
escrever_resultado(caminho_saida, ator, media)


```

```python
def contar_aparicoes_filmes(linhas):
    contagem_filmes = {}
    
    for linha in linhas[1:]: 
        colunas = linha.strip().split(',')
        
        if len(colunas) < 5:  
            continue
        
        filme = colunas[4].strip()  
        
        if filme in contagem_filmes:
            contagem_filmes[filme] += 1
        else:
            contagem_filmes[filme] = 1
    
  
    filmes_ordenados = sorted(contagem_filmes.items(), key=lambda x: (-x[1], x[0]))
    
    return filmes_ordenados


def escrever_resultado(filepath, filmes_ordenados):
    with open(filepath, 'w', encoding='utf-8') as file:
        for filme, quantidade in filmes_ordenados:
            file.write(f"O filme {filme} aparece {quantidade} vez(es) no dataset\n")


caminho_csv = 'actors.csv'
caminho_saida = 'etapa-4.txt'


linhas = ler_arquivo(caminho_csv)
filmes_ordenados = contar_aparicoes_filmes(linhas)
escrever_resultado(caminho_saida, filmes_ordenados)



```

```python
def ordenar_atores_por_receita(linhas):
    atores_receita = []
    
    for linha in linhas[1:]:  
        colunas = linha.strip().split(',')
        
        if len(colunas) < 2:  
            continue
        
        ator = colunas[0]
        
        try:
            receita_str = colunas[1].strip().replace(',', '.')  
            receita = float(receita_str)  
            if receita < 0:  
                continue
        except ValueError:
            continue  
        
        atores_receita.append((ator, receita))
    
   
    atores_ordenados = sorted(atores_receita, key=lambda x: -x[1])
    
    return atores_ordenados


def escrever_resultado(filepath, atores_ordenados):
    with open(filepath, 'w', encoding='utf-8') as file:
        for ator, receita in atores_ordenados:
            file.write(f"{ator} - {receita:.2f} milhões de dólares\n")


caminho_csv = 'actors.csv'
caminho_saida = 'etapa-5.txt'


linhas = ler_arquivo(caminho_csv)
atores_ordenados = ordenar_atores_por_receita(linhas)
escrever_resultado(caminho_saida, atores_ordenados)

```
